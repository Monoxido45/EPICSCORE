# Printing and obtaining additional results for quantile regression and regression:

```python
import os
import pandas as pd

original_path = os.getcwd()

def custom_table_generator(result):
    metrics = list(result.columns)[1:]

    print(f"\\begin{{tabular}}{{l{'c' * len(metrics)}}}\\toprule")
    print("Dataset & " + " & ".join(metrics) + "\\\\\\midrule")
    for _, row in result.iterrows():
        values = [row["Dataset"]]
        for metric in result.columns[1:]:
            value = row[metric]
            if pd.isna(value):
                values.append("-")
            else:
                values.append(f"{value}")
        print(" & ".join(values) + "\\\\")
    print("\\bottomrule\\end{tabular}")
```

Getting the results:

```python
folder_path = os.path.dirname(original_path) + "/Experiments_code/results"
```

## Quantile regression results:

Reading pickle files and transforming to table:
* Interval length ratio between outlier and inliers:

```python
output_ratio = pd.read_pickle(folder_path + "/result_ratio_outlier.pkl")
output_ratio
```

```python
custom_table_generator(output_ratio)
```

* AISL per outlier:

```python
output_aisl = pd.read_pickle(folder_path + "/result_aisl_outlier.pkl")
output_aisl
```

```python
custom_table_generator(output_aisl)
```

* coverage per outlier:

```python
output_cover = pd.read_pickle(folder_path + "/result_cover_outlier.pkl")
output_cover
```

```python
custom_table_generator(output_cover)
```
